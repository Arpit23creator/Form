```python
Q1 = df["grades"].quantile(0.25)
Q3 = df["grades"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("lower limit:", lower)
print("upper limit:", upper)
```
